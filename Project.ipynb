```python

import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display

print("Bridge Load Distribution Calculator")
print("Libraries loaded successfully!")
```

```python

def calculate_bridge(P1, P2):
    # Matrix equation AX = B
    A = np.array([[1, 1],
                  [0, 10]], dtype=float)

    B = np.array([P1 + P2,
                  3 * P1 + 7 * P2], dtype=float)

    # Solve using the matrix inverse
    X = np.linalg.inv(A) @ B

    RA = X[0]
    RB = X[1]

    # Display mathematical calculation
    print("MATRIX A:")
    print(A)

    print("\nMATRIX B:")
    print(B)

    print("\nREACTIONS: X = inverse(A) x B")
    print(f"Reaction at A = {RA:.2f} kN")
    print(f"Reaction at B = {RB:.2f} kN")

    print("\nCHECK:")
    print(f"RA + RB = {RA + RB:.2f} kN")
    print(f"P1 + P2 = {P1 + P2:.2f} kN")

    # Draw bridge diagram and reactions
    fig, ax = plt.subplots(1, 2, figsize=(12, 4))

    ax[0].plot([0, 10], [0, 0], "k-", linewidth=5)
    ax[0].scatter([0, 10], [0, 0],
                  color="green", s=100, zorder=3)

    # Downward load arrows
    for x, p, label in [(3, P1, "P1"), (7, P2, "P2")]:
        ax[0].annotate(
            f"{label} = {p} kN",
            xy=(x, 0), xytext=(x, 1.8),
            ha="center",
            arrowprops=dict(arrowstyle="->", color="red"),
            color="red"
        )

    # Upward reaction arrows
    for x, r, label in [(0, RA, "RA"), (10, RB, "RB")]:
        ax[0].annotate(
            f"{label} = {r:.2f} kN",
            xy=(x, 0), xytext=(x, -1.3),
            ha="center",
            arrowprops=dict(arrowstyle="->", color="blue"),
            color="blue"
        )

    ax[0].set_xlim(-1, 11)
    ax[0].set_ylim(-2, 2.5)
    ax[0].set_title("Bridge and Support Reactions")
    ax[0].set_xlabel("Distance along bridge (m)")
    ax[0].set_yticks([])
    ax[0].grid(axis="x", alpha=0.3)

    # Bar chart
    ax[1].bar(["Support A", "Support B"], [RA, RB])
    ax[1].set_title("Support Reaction Comparison")
    ax[1].set_ylabel("Reaction force (kN)")
    ax[1].grid(axis="y", alpha=0.3)

    plt.tight_layout()
    plt.show()


# Interactive load sliders
widgets.interact(
    calculate_bridge,
    P1=widgets.IntSlider(
        value=10, min=0, max=100, step=5,
        description="Load P1 (kN)"
    ),
    P2=widgets.IntSlider(
        value=20, min=0, max=100, step=5,
        description="Load P2 (kN)"
    )
);
```
